## Introduction

This project analyses 14 days of sales and stock records for Wanjiku, a fresh-produce stall owner at Gikomba Market. The objective is to evaluate business performance, identify sales trends, compare weekday and weekend sales, assess stock levels, calculate product gross margins, and provide practical recommendations to improve business operations.


```python

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```


## Task 1: Overall Sales Performance

In this task, I will use Python's built-in functions to calculate total sales, average daily sales, the highest sales and the lowest sales over the 14-day period. I will also use list indexing to identify the day and week when the best and worst sales occurred.


```python

# Calculate overall sales statistics

total_sales = sum(daily_sales)

average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)

lowest_sales = min(daily_sales)

print("Total Sales:", total_sales)
print("Average Daily Sales:", average_sales)
print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)
```

```python

# Find the positions of the highest and lowest sales

best_index = daily_sales.index(highest_sales)

worst_index = daily_sales.index(lowest_sales)


# Identify the corresponding days

best_day = days[best_index]

worst_day = days[worst_index]


# Identify the weeks

if best_index < 7:
    best_week = 1
else:
    best_week = 2

if worst_index < 7:
    worst_week = 1
else:
    worst_week = 2


# Display the results

print("Best Sales Day:", best_day)
print("Best Sales Week:", best_week)

print("Worst Sales Day:", worst_day)
print("Worst Sales Week:", worst_week)
```


## Task 2: Weekly Sales Totals and Growth

In this task, I will define a reusable function called week_totals() to calculate the total sales for Week 1 and Week 2. I will use list slicing to separate the 14 days into two weeks, return the weekly totals as a tuple, and calculate the percentage growth between the two weeks.


```python

def week_totals(sales):
    """
    Calculate total sales for Week 1 and Week 2.

    Args:
        sales: A list containing 14 daily sales figures.

    Returns:
        A tuple containing the Week 1 and Week 2 sales totals.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])

    return week1, week2
```

```python

# Call the function and unpack the results

week1_total, week2_total = week_totals(daily_sales)

print("Week 1 Total:", week1_total)
print("Week 2 Total:", week2_total)
```

```python

# Calculate week-on-week growth percentage

growth_percentage = ((week2_total - week1_total) / week1_total) * 100

print("Week-on-Week Growth:", growth_percentage, "%")
```


## Task 3: Daily Sales Classification

In this task, I will classify each day's sales as Strong, Normal or Weak by comparing daily sales against the average daily sales. I will define a reusable function called classify_day() using conditional statements, apply it to all 14 days using a for loop, and count the number of days in each category using a dictionary.


```python

def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal or Weak.

    Args:
        amount: The sales amount for a particular day.
        average: The average daily sales.

    Returns:
        Strong, Normal or Weak depending on sales performance.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"
```

```python

print(classify_day(5000, average_sales))
print(classify_day(3000, average_sales))
print(classify_day(2000, average_sales))
```

```python

# Create an empty list to store daily classifications

daily_classifications = []

# Loop through every day's sales

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)

print(daily_classifications)
```

```python

# Create a dictionary to count each category

category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Count the classifications

for category in daily_classifications:
    category_counts[category] += 1

print(category_counts)
```

```python

print("DAILY SALES CLASSIFICATION")

for i in range(len(daily_sales)):
    print(days[i], "-", daily_sales[i], "-", daily_classifications[i])

print("\nCATEGORY SUMMARY")

for category, count in category_counts.items():
    print(category, ":", count, "days")
```

## Task 4: Weekday vs Weekend Performance

This task compares sales performance on weekdays (Monday to Friday) and weekends (Saturday and Sunday).

The objective is to determine whether the market generates higher average sales during weekdays or weekends.

```python
# Create empty lists to store weekday and weekend sales
weekday_sales = []
weekend_sales = []

# Separate sales based on the day of the week
for day, sale in zip(days, daily_sales):
    if day in ["Sat", "Sun"]:
        weekend_sales.append(sale)
    else:
        weekday_sales.append(sale)

print("Weekday Sales:", weekday_sales)
print("Weekend Sales:", weekend_sales)
```

```python
# Calculate total sales
weekday_total = sum(weekday_sales)
weekend_total = sum(weekend_sales)

# Calculate average sales
weekday_average = weekday_total / len(weekday_sales)
weekend_average = weekend_total / len(weekend_sales)

print("Total Weekday Sales:", weekday_total)
print("Total Weekend Sales:", weekend_total)

print("Average Weekday Sales:", weekday_average)
print("Average Weekend Sales:", weekend_average)
```

```python
if weekday_average > weekend_average:
    print("Weekdays have higher average sales.")
elif weekend_average > weekday_average:
    print("Weekends have higher average sales.")
else:
    print("Weekdays and weekends have equal average sales.")
```

### Business Insight

Weekend sales have a higher average of ****KES** 4,**900**** compared to weekday average sales of ****KES** 2,**685****.

This represents approximately **82.50% higher average sales during weekends** compared to weekdays.

Although total weekday sales may be higher because there are more weekdays in the dataset, weekend days generate higher average sales.

**Recommendation:** Wanjiku should consider increasing stock availability and preparing adequately for weekend demand to take advantage of the higher average sales. She should also monitor weekday sales and consider promotions to attract more customers during slower periods.

## Task 5: Stock Analysis

This task analyses the current stock levels of products in the Gikomba Market stall.

The objective is to identify products with low stock, calculate the total available inventory, and determine which products require restocking.

```python

def get_reorder_items(stock):
    """
    Return products whose quantity is at or below their reorder level.

    Args:
        stock: Dictionary containing product stock information.

    Returns:
        List of product names that need restocking.
    """
    reorder_items = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_items.append(product)

    return reorder_items


```

```python

reorder_items = get_reorder_items(stock)

print("Products that need restocking:")
print(reorder_items)
```

### Business Insights

**Stock Replenishment**

Tomatoes and Sukuma need restocking because their available quantities are below their reorder levels.

**Gross Profit Margin**

Avocado has the highest gross margin at 42.86%, followed by Sukuma at 40%.

Onions have the lowest gross margin at 30%.

**Recommendations**

1. Restock Tomatoes and Sukuma to avoid running out of stock.
2. Monitor Avocado inventory because it has the highest gross profit margin.
3. Review the pricing and purchasing costs of Onions to identify opportunities to improve profitability.
4. Regularly monitor inventory levels to maintain sufficient stock.

## Task 6: Stall Performance Report

This task brings together the findings from the previous tasks to produce a formatted Stall Performance Report.

```python
# Create a list of products that need restocking
low_stock_products = []

for product, details in stock.items():
    if details["qty"] <= details["reorder"]:
        low_stock_products.append(product)

print("Products requiring restocking:", low_stock_products)
```

```python


def print_stall_report():
    """
    Print a formatted summary of Wanjiku's stall performance.

    The report includes sales performance, weekly growth,
    weekday and weekend averages, stock levels, profit
    margins, and products requiring restocking.
    """

    print("=" * 55)
    print("       GIKOMBA MARKET STALL PERFORMANCE REPORT")
    print("=" * 55)

    # Sales performance
    print("\nSALES PERFORMANCE")
    print("-" * 55)

    print(f"Total Sales:              KES {total_sales:,.2f}")
    print(f"Average Daily Sales:      KES {average_daily_sales:,.2f}")
    print(f"Week 1 Sales:             KES {week_1_sales:,.2f}")
    print(f"Week 2 Sales:             KES {week_2_sales:,.2f}")
    print(f"Weekly Sales Growth:      {weekly_growth:.2f}%")

    # Weekday and weekend performance
    print("\nWEEKDAY VS WEEKEND PERFORMANCE")
    print("-" * 55)

    print(f"Weekday Average:          KES {weekday_average:,.2f}")
    print(f"Weekend Average:          KES {weekend_average:,.2f}")

    # Stock performance and profit margins
    print("\nSTOCK PERFORMANCE")
    print("-" * 55)

    for product, details in stock.items():

        margin = (
            (details["price"] - details["cost"])
            / details["price"]
        ) * 100

        print(
            f"{product:<12} "
            f"Quantity: {details['qty']:<4} "
            f"Margin: {margin:.2f}%"
        )

    # Products requiring restocking
    print("\nPRODUCTS REQUIRING RESTOCKING")
    print("-" * 55)

    reorder_items = get_reorder_items(stock)

    for product in reorder_items:
        print(f"- {product}")

    # Business recommendations
    print("\nRecommendations for Wanjiku")
    print("-" * 55)

    print("1. Restock Tomatoes and Sukuma to prevent shortages.")
    print("2. Prepare additional stock for higher weekend demand.")
    print("3. Monitor weekly sales growth and business trends.")
    print("4. Focus on products with higher profit margins.")

    print("=" * 55)
    print("                 END OF REPORT")
    print("=" * 55)
```
